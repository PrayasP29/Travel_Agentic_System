```python
import httpx

urls = [
    "https://mcp.kiwi.com",
    "https://gribstream.com/mcp",
    "https://mcp.agentorist.com/mcp"
]

for url in urls:
    try:
        response = httpx.get(url, timeout=10)
        print(url)
        print("Status:", response.status_code)
        print("-" * 40)
    except Exception as e:
        print(url)
        print("ERROR:", e)
        print("-" * 40)
```
